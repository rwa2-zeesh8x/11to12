```python
import numpy as np

x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 5, 4, 6], dtype=float)

# b = (nΣxy - Σx·Σy) / (nΣx² - (Σx)²)
b = (len(x) * np.sum(x * y) - np.sum(x) * np.sum(y)) / \
    (len(x) * np.sum(x**2) - np.sum(x)**2)

# a = ȳ - b·x̄
a = np.mean(y) - b * np.mean(x)

print(f"y = {a:.2f} + {b:.2f}x")   
```
